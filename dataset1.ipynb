```python
import os
import glob
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

DATA_DIR = "C:\\Users\\kanam\\OneDrive\\Documents\\project\\PhysioNet2019\\training_setA";

files = glob.glob(os.path.join(DATA_DIR, "**", "*.psv"), recursive=True)

print("Total patient files:", len(files))

if not files:
    raise FileNotFoundError(
        f"No .psv files found in {DATA_DIR}. Check your dataset path."
    )

dataframes = []

for file in files:
    df = pd.read_csv(file, sep="|")
    df["Patient_ID"] = os.path.basename(file).replace(".psv", "")
    df["Hour"] = np.arange(len(df))
    dataframes.append(df)

data = pd.concat(dataframes, ignore_index=True)

print("Dataset shape:", data.shape)
print(data.head())
```

```python
print(data.columns.tolist())
```

```python
missing_count = data.isnull().sum()

missing_percentage = (data.isnull().sum() / len(data)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Sort by highest missing percentage
missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

print(missing_summary.to_string())
```

```python
# Create a copy of the original dataset
data_clean = data.copy()

# Identify columns with 100% missing values
all_missing_cols = [
    col for col in data_clean.columns
    if data_clean[col].isnull().all()
]

print("Columns with 100% missing values:", all_missing_cols)

# Remove columns that contain no recorded values
data_clean = data_clean.drop(columns=all_missing_cols)

print("\nOriginal dataset shape:", data.shape)
print("Shape after removing completely missing columns:", data_clean.shape)

print("\nRemaining columns:")
print(data_clean.columns.tolist())
```

```python
print(data_clean.isnull().sum())
```

```python
data_clean = data_clean.drop_duplicates()

print("Dataset shape after removing duplicates:", data_clean.shape)
print("Remaining duplicate records:", data_clean.duplicated().sum())
```

```python
from sklearn.impute import SimpleImputer

# Sort records by patient and hour
data_clean = data_clean.sort_values(
    ["Patient_ID", "Hour"]
).copy()

# Select feature columns
exclude_cols = ["SepsisLabel", "Patient_ID", "Hour"]
feature_cols = [
    col for col in data_clean.columns
    if col not in exclude_cols
]

# Step 1: Forward fill within each patient
data_clean[feature_cols] = (
    data_clean.groupby("Patient_ID")[feature_cols].ffill()
)

# Step 2: Median imputation for remaining missing values
imputer = SimpleImputer(strategy="median")

data_clean[feature_cols] = imputer.fit_transform(
    data_clean[feature_cols]
)

# Check remaining missing values
print("Missing values after imputation:")
print(data_clean[feature_cols].isnull().sum())

print("\nTotal missing feature values:",
      data_clean[feature_cols].isnull().sum().sum())
```

```python
print(data_clean["SepsisLabel"].value_counts())
```

```python
# Target variable
y = data_clean["SepsisLabel"]

# Input features: all columns except target and identifiers
X = data_clean.drop(columns=["SepsisLabel", "Patient_ID", "Hour"])

print("Input features:", X.columns.tolist())
print("Target column:", y.name)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import GroupShuffleSplit

groups = data_clean["Patient_ID"]

# First split: 80% training, 20% remaining
split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, temp_idx = next(
    split1.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
y_train = y.iloc[train_idx].copy()

X_temp = X.iloc[temp_idx].copy()
y_temp = y.iloc[temp_idx].copy()

groups_temp = groups.iloc[temp_idx]

# Second split: divide remaining 20% equally
split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(
    split2.split(X_temp, y_temp, groups=groups_temp)
)

X_val = X_temp.iloc[val_idx].copy()
y_val = y_temp.iloc[val_idx].copy()

X_test = X_temp.iloc[test_idx].copy()
y_test = y_temp.iloc[test_idx].copy()

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaler to validation and testing data
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

```

```python
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("scale_pos_weight:", scale_pos_weight)
```

```python
from xgboost import XGBClassifier
import joblib

xgb_model = XGBClassifier(
    n_estimators=1000,
    max_depth=6,
    learning_rate=0.05,
    scale_pos_weight=scale_pos_weight,
    eval_metric="aucpr",
    early_stopping_rounds=30,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=False
)

joblib.dump(xgb_model, "xgboost_sepsis_model.pkl")

# print("Best validation PR-AUC:", xgb_model.best_score)
```

```python
y_val_prob = xgb_model.predict_proba(X_val)[:, 1]
y_val_pred = (y_val_prob >= 0.5).astype(int)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Classification Report:")
print(classification_report(y_val, y_val_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred))

print("ROC-AUC:", roc_auc_score(y_val, y_val_prob))
print("PR-AUC:", average_precision_score(y_val, y_val_prob))
```
